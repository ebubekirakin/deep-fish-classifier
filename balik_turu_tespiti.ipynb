### GPU ve çalışma aygıtı kontrolü
Aşağıdaki hücre GPU'nun kullanılabilir olup olmadığını kontrol eder. GPU yoksa proje CPU'da da çalışabilir; ancak eğitim daha yavaş olabilir.

### Tekrarlanabilirlik için seed
Seed, rastgele seçimlerin başlangıç değeridir. Aynı seed kullanıldığında sonuçların benzer şekilde tekrarlanması kolaylaşır; tüm donanımlarda sonuçların tamamen aynı olacağı garanti edilmez.

```python
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)
print(f"Seed ayarlandı: {SEED}")
```

Learning rate, modelin her öğrenme adımında ağırlıklarını ne kadar değiştireceğini belirler.

Bu değerler ileride kolayca değiştirilebilsin diye tek hücrede tutulmuştur.
